## Get tar file

```python
# Paths
save_to_path = "/content/drive/My Drive/Thesis/images_dataset/"
tar_path = "/content/origin.tar"
tar_url = "https://www.cs.rice.edu/~vo9/visualnews/origin.tar"

# Step 1: Download the tar file (if not already downloaded)
if not os.path.exists(tar_path):
    !wget {tar_url} -P /content/
```

## download relevant images

```python
# Step 2: Load the list of image paths
# tobedownloaded = pd.read_csv('your_list.csv')  # if needed
image_paths_to_extract = set(tobedownloaded['image_path_origin_format'].tolist())

# Step 3: Extract files and print progress
with tarfile.open(tar_path, 'r') as tar:
    count = 0
    for member in tar.getmembers():
        if member.name in image_paths_to_extract:
            dest_path = os.path.join(save_to_path, member.name)
            if not os.path.exists(dest_path):
                # Ensure target directories exist
                os.makedirs(os.path.dirname(dest_path), exist_ok=True)

                # Extract and save the file
                with open(dest_path, "wb") as f_out:
                    f_out.write(tar.extractfile(member).read())

                count += 1
                print(count, member.name)

print(f"\n✅ Done! Extracted {count} new files to {save_to_path}")
```
