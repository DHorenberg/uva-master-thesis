```python
#load data
df_responses = pd.read_csv('/content/drive/MyDrive/Thesis/analysis/human_feedback/human_feedback_survey.csv')
```

```python
columns_to_replace = df_responses.columns.difference(['Respondent'])

# Replace 1 and 2 with Foto 1 and Foto 2 so that they can match
df_responses[columns_to_replace] = df_responses[columns_to_replace].replace(
    {"1": "Foto 1", "2": "Foto 2", 1: "Foto 1", 2: "Foto 2"}
)
```

```python
# expected answers for lowest iconcity images
expected_answers = {
    'Q1': 'Foto 1', 'Q49': 'Foto 1', 'Q50': 'Foto 1', 'Q51': 'Foto 1', 'Q52': 'Foto 2', 'Q53': 'Foto 1',
    'Q54': 'Foto 2', 'Q55': 'Foto 2', 'Q56': 'Foto 2', 'Q57': 'Foto 2', 'Q58': 'Foto 1', 'Q59': 'Foto 2',
    'Q60': 'Foto 2', 'Q61': 'Foto 1', 'Q62': 'Foto 1', 'Q63': 'Foto 1', 'Q64': 'Foto 1', 'Q65': 'Foto 2',
    'Q66': 'Foto 2', 'Q67': 'Foto 2', 'Q68': 'Foto 1', 'Q69': 'Foto 1', 'Q70': 'Foto 2', 'Q71': 'Foto 2',
    'Q72': 'Foto 1', 'Q73': 'Foto 2', 'Q74': 'Foto 1', 'Q75': 'Foto 2', 'Q76': 'Foto 2', 'Q77': 'Foto 2',
    'Q89': 'Foto 2', 'Q90': 'Foto 2', 'Q91': 'Foto 2', 'Q92': 'Foto 1', 'Q93': 'Foto 2', 'Q94': 'Foto 1',
    'Q95': 'Foto 1', 'Q96': 'Foto 1', 'Q97': 'Foto 2', 'Q98': 'Foto 1'
}

# check for matches
df_long = df_responses.melt(id_vars='Respondent', var_name='Question', value_name='Answer')
df_long['Expected'] = df_long['Question'].map(expected_answers)
df_long['IsMatch'] = df_long['Answer'] == df_long['Expected']

# calcualte consistency
overall_consistency = df_long['IsMatch'].mean()
print(f"Overall Consistency: {overall_consistency:.2%}")

# consistency for each question
question_consistency = df_long.groupby('Question')['IsMatch'].mean().reset_index()
print(question_consistency)


n = len(df_long)
k = df_long['IsMatch'].sum()
p_val = binomtest(k, n, p=0.5, alternative='greater')
print(f"Binomial test p-value: {p_val}")
```

```python

from scipy.stats import binomtest
import pandas as pd

# link politician to question
subject_politician = {
    'Q1': 'trump', 'Q49': 'trump', 'Q50': 'trump', 'Q51': 'trump', 'Q52': 'trump', 'Q53': 'trump',
    'Q54': 'trump', 'Q55': 'trump', 'Q56': 'trump', 'Q57': 'trump', 'Q58': 'merkel', 'Q59': 'merkel',
    'Q60': 'merkel', 'Q61': 'merkel', 'Q62': 'merkel', 'Q63': 'merkel', 'Q64': 'merkel', 'Q65': 'merkel',
    'Q66': 'merkel', 'Q67': 'merkel', 'Q68': 'obama', 'Q69': 'obama', 'Q70': 'obama', 'Q71': 'obama',
    'Q72': 'obama', 'Q73': 'obama', 'Q74': 'obama', 'Q75': 'obama', 'Q76': 'obama', 'Q77': 'obama',
    'Q89': 'rutte', 'Q90': 'rutte', 'Q91': 'rutte', 'Q92': 'rutte', 'Q93': 'rutte', 'Q94': 'rutte',
    'Q95': 'rutte', 'Q96': 'rutte', 'Q97': 'rutte', 'Q98': 'rutte'
}
df_long['Subject'] = df_long['Question'].map(subject_politician)
results = []
for subject in df_long['Subject'].unique():
    subset = df_long[df_long['Subject'] == subject]
    n = len(subset)
    k = subset['IsMatch'].sum()
    p_val = binomtest(k, n, p=0.5, alternative='greater').pvalue
    results.append({'Subject': subject, 'N': n, 'Matches': k, 'Consistency': k/n, 'P-value': p_val})

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(by='Consistency', ascending=False)

print(results_df)
```
